```python
import pandas as pd

# Criando um DataFrame
data = {
    'Nome': ['Alice', 'Bob', 'Charlie'],
    'Idade': [25, 30, 35],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
print("\n")
# Adicionando uma nova coluna
df['Salário'] = [50000, 60000, 70000]
print("DataFrame após adicionar nova coluna:")
print(df)
print("\n")
# Filtrando dados
filtro = df['Idade'] > 28
df_filtrado = df[filtro]    
```

```python
# Criando um novo DataFrame
novos_dados = {
    'Nome': ['Cleyber', 'Juciara', 'Lara', 'Júlia'],
    'Idade': [46, 45, 20, 17]
}
df_novo = pd.DataFrame(novos_dados)
print("Novo DataFrame:")
print(df_novo)
```

```python
# Adicionando a coluna "Ano de Nascimento"
df_novo['Ano de Nascimento'] = [1978, 1979, 2004, 2007]
print("DataFrame após adicionar a coluna 'Ano de Nascimento':")
print(df_novo)
```

```python
# Atualizando a coluna 'Nome' com os nomes completos
df_novo['Nome'] = [
    'Cleyber Gomes da Silva',
    'Juciara Francine Gomes da Silva',
    'Lara Caylane Gomes da Silva',
    'Júlia Kimberly Gomes da Silva'
]
print("DataFrame após atualizar os nomes:")
print(df_novo)
```
